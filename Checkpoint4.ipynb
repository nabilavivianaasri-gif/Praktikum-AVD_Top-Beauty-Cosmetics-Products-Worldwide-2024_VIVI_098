```python
results = []

cols = filenyanih.select_dtypes(include=['float64', 'int64'])

for col in cols:
    q1 = filenyanih[col].quantile(0.25)
    q3 = filenyanih[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = filenyanih[(filenyanih[col] < lower_bound) |
                           (filenyanih[col] > upper_bound)]

    percent_outliers = (len(outliers) / len(filenyanih)) * 100

    results.append({'Kolom': col,'Persentase Outliers (%)': percent_outliers})

results_filenyanih = pd.DataFrame(results)
results_filenyanih.set_index('Kolom', inplace=True)

display(results_filenyanih)
```

# **Construct Data**

```python
def VARIABEL(Popularity_Score):
    if Popularity_Score > 20000:
        return 'High'
    elif Popularity_Score < 10000:
        return 'Medium'
    else:
        return 'Low'

filenyanih['Popularity_Score'] = filenyanih['Rating'] * filenyanih['Number_of_Reviews']
filenyanih['Tipe'] = filenyanih['Popularity_Score'].apply(VARIABEL)
```

```python
print(filenyanih.columns)
```

Penambahan kolom baru dilakukan untuk menghasilkan informasi yang lebih representatif dibandingkan hanya menggunakan data mentah seperti Rating dan Number_of_Reviews secara terpisah. Kolom Popularity_Score dibuat untuk menggabungkan kedua variabel tersebut agar dapat mencerminkan tingkat popularitas produk secara lebih akurat.

Selain itu, pengelompokan ke dalam kategori seperti High, Medium, dan Low mempermudah interpretasi data dan membantu perusahaan dalam menentukan strategi pemasaran serta pengambilan keputusan bisnis
